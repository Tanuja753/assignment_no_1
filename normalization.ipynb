```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
df=pd.read_csv("class_ipl_22-26_dataset.csv")
df.head()
```

```python
x=df[["team1_runs"]]
x.head()
```

```python
x.index
```

```python
scaler=MinMaxScaler()
x_scaled=scaler.fit_transform(x)
x_scaled=pd.DataFrame(x_scaled,columns=["team1_runs"])
x_scaled.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sb
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.title("team1 runs before scaling")
sb.scatterplot(x=x.index,y=x['team1_runs'])
plt.subplot(1,2,2)
plt.title("team1 runs after scaling")
sb.scatterplot(x=x_scaled.index,y=x_scaled['team1_runs'])
plt.show()
```

```python
print("original data: ",x.describe())
print("scaled data: ",x_scaled.describe())
```
