```python
import pandas as pd
import matplotlib.pyplot as plt
ipl=pd.read_csv("IPLPlayerStat.csv")
ipl.std(numeric_only=True)
batting_avg_std=ipl["batting_avg"].std()
plt.figure(figsize=(12,6))
ipl["batting_avg"].plot(kind="hist",bins=15,edgecolor="black",alpha=0.7,density=True)
ipl["batting_avg"].plot(kind="kde",color="red",ax=plt.gca(),secondary_y=False)
plt.title("distribution of batting average")
plt.xlabel("batting average")
plt.ylabel('Density/Frequency')
plt.xticks(rotation=45,ha='right')
plt.tight_layout()
plt.show()
```
