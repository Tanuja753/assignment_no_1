```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
df=pd.read_csv("class_ipl_22-26_dataset.csv")
df.head()
```

```python
df[["team1_runs"]].head(10)
```

```python
scaler=StandardScaler()
df["team1_runs_scaled"]=scaler.fit_transform(df[["team1_runs"]])
df[["team1_runs","team1_runs_scaled"]].head(20)
```

```python
df["team1_wickets_scaled"]=scaler.fit_transform(df[["team1_wickets"]])
df[["team1_wickets","team1_wickets_scaled"]].head(20)
```

```python
df[["team1_runs","team1_runs_scaled"]].describe().round(2)
```

```python
df[["team1_wickets","team1_wickets_scaled"]].describe().round(2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sb
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.title("team1 runs before scaling")
sb.histplot(df["team1_runs"],bins=10,kde=True)
plt.subplot(1,2,2)
plt.title("team1 runs after scaling")
sb.histplot(df["team1_runs_scaled"],bins=10,kde=True)
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.title("team1 wickets before scaling")
sb.histplot(df["team1_wickets"],bins=10,kde=True)
plt.subplot(1,2,2)
plt.title("team1 wickets after scaling")
sb.histplot(df["team1_wickets_scaled"],bins=10,kde=True)
plt.show()
```
