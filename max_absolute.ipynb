```python
import pandas as pd
from sklearn.preprocessing import MaxAbsScaler
df=pd.read_csv("class_ipl_22-26_dataset.csv")
df.head()
```

```python
x=df[['team1_runs']]
x.head()
```

```python
scaler=MaxAbsScaler()
x_scaled=scaler.fit_transform(x)
x_scaled=pd.DataFrame(x_scaled,columns=["team1_runs"])
x_scaled.head()
```

```python
comparison=pd.DataFrame({
    "original":x["team1_runs"],
    "scaled":x_scaled["team1_runs"]
})
#pd.set_option('display.max_rows', None)
comparison.head()
```

```python
print("original data: ",x.describe())
print("scaled data: ",x_scaled.describe())
```
