```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
df=pd.read_csv("class_ipl_22-26_dataset.csv")
df.head()
```

```python
df[["winner"]].head()
```

```python
print("missing values: ",df["winner"].isnull().sum())
print("unique winners: ",df["winner"].nunique())
```

```python
le=LabelEncoder()
df["encoded_winner"]=le.fit_transform(df["winner"])
df[["winner","encoded_winner"]].head(11)
```

```python
mapping=pd.DataFrame({
    'winner':le.classes_,
    'encoded_winner':range(len(le.classes_)),
})
mapping
```
