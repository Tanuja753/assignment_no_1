```python
virat=np.array([10,62,76,86,54,86,97,98,57,76])
rohit=np.array([98,65,53,86,32,21,54,76,43,65])
print(np.sum(virat))
print(np.sum(rohit))
if np.sum(virat) > np.sum(rohit):
    print("virat scored more runs")
else:
    print("rohit scored more runs")
print("difference : ",np.mean(virat) - np.mean(rohit))
count=0
for i,j in zip(virat,rohit):
    if i >j:
        count+=1
    i=i+1
    j=j+1
print("count: ",count)
```

```python
team_runs=np.array([[45,60,30],[50,70,40],[20,30,50],[60,40,80],[30,20,40]])
print(np.sum(team_runs,axis=1))
print(np.sum(team_runs,axis=0))
print(np.max(np.sum(team_runs,axis=1)))
```

```python
player=np.array([10,62,76,86,54,86,97,98,57,76])
print(np.mean(player))
print(np.std(player))
print(np.var(player))
if np.var(player)<15:
    print("player is consistent")
else:
    print("player is not consistent")
```

```python
runs=np.array([1,4,0,6,2,1,4,0,1,6,2,1])
print("total runs : ",np.sum(runs))
fours=0
sixs=0
for i in runs:
    if i ==6:
        sixs=sixs+1
    elif i==4:
        fours=fours+1
    i=i+1
print("fours : ",fours)
print("sixs : ",sixs)
print("total boundry runs : ",(fours*4)+(sixs*6))
print(np.size(runs))
print("percentage of runs scored through boundries :",((fours+sixs)/np.size(runs))*100)
```
