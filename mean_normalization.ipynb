```python
import pandas as pd
df=pd.read_csv("class_ipl_22-26_dataset.csv")
df.head()
```

```python
x=df[['team1_runs']]
x.head()
```

```python
mean=x['team1_runs'].mean()
minimum=x['team1_runs'].min()
maximum=x['team1_runs'].max()
print("mean:",mean)
print("minimum:",minimum)
print("maximum:",maximum)
```

```python
x_mean=x.copy()
x_mean['team1_runs']=(x_mean['team1_runs']-mean)/(maximum-minimum)
x_mean.head()
```

```python
comparison=pd.DataFrame({
    'original':x['team1_runs'],
    'mean_scaled':x_mean['team1_runs']
})
#pd.set_option('display.max_rows', None)
comparison.head()
```

```python
print("original data: ",x.describe().round(2))
print("mean normalized: ",x_mean.describe().round(2))
```
