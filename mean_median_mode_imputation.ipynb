```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv("class_ipl_22-26_dataset_with_nulls.csv")
print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
mean_df=df.copy()
mean_value=mean_df["team1_wickets"].mean()
print("mean_value",mean_value)
mean_df["team1_wickets"]=mean_df["team1_wickets"].fillna(mean_value)
mean_df["team1_wickets"].isnull().sum()
```

```python
mean_df.isnull().sum()
```

```python
median_df=df.copy()
print(median_df["team1_wickets"].isnull().sum())
median_value=median_df["team1_wickets"].median()
print("median_value",median_value)
median_df["team1_wickets"]=median_df["team1_wickets"].fillna(median_value)
print(median_df["team1_wickets"].isnull().sum())
```

```python
mode_df=df.copy()
print(mode_df["winner"].isnull().sum())
mode_value=mode_df["winner"].mode()
print("mode_value",mode_value)
mode_df["winner"]=mode_df["winner"].fillna(mode_value[0])
print(mode_df["winner"].isnull().sum())
```

using simple imputer

```python
df.isnull().sum()
```

```python
print(df.isnull().sum())
mean_imp=SimpleImputer(strategy="mean")
median_imp=SimpleImputer(strategy="median")
mode_imp=SimpleImputer(strategy="most_frequent")
df["team1_wickets"]=mean_imp.fit_transform(df[["team1_wickets"]])
df[["winner"]]=mode_imp.fit_transform(df[["winner"]])
df[["player_of_match"]]=mode_imp.fit_transform(df[["player_of_match"]])
print(df.isnull().sum())
```
