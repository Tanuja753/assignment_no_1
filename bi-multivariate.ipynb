```python
import pandas as pd
import  matplotlib.pyplot as plt
import seaborn as sb
df=pd.read_csv("class_ipl_22-26_dataset.csv")
df.head()
```

```python
df.shape
df.columns
```

```python
sb.scatterplot(data=df.tail(74),x='team1_wickets',y='team1_runs')
```

```python
sb.scatterplot(data=df.tail(74),x='team1',y='team1_runs',hue='winner').tick_params(axis='x',rotation=90)
```

```python
multi_var_cols=['team1_runs','team2_runs','win_by_runs','win_by_wickets','season','result_type']
multi_var_df=df[multi_var_cols].head(100)
sb.pairplot(data=multi_var_df,hue='result_type',diag_kind='kde')
plt.show()
```

```python
numerical_cols=['season','win_by_runs','win_by_wickets','team1_runs']
cor_df=df[numerical_cols]
cor=cor_df.corr()
sb.heatmap(cor,annot=True,cmap=plt.cm.Blues)
plt.show()
```
