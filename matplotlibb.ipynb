```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("class_ipl_22-26_dataset.csv")
df
```

```python
df.info()
```

```python
print(df['team2_wickets'])
```

```python
plt.plot(df['team1_runs'].tail(74))
plt.xlabel('match index')
plt.ylabel('team 1 runs')
plt.title('team1 runs trend')
plt.show()
```

```python
plt.plot(df['team2_runs'].tail(74),'g--',label='team2 runs')
plt.xlabel('match index')
plt.ylabel('team 2 runs')
plt.title('team2 runs trend')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df['team1_runs'].tail(74),color='blue',label='team1 runs')
plt.plot(df['team1_wickets'].tail(74),color='red',label='team1 wickets')
plt.xlabel('wickets')
plt.ylabel('runs')
plt.title('team 1 runs vs wickets')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df['team1_runs'].tail(74),color='blue',label='team1 runs')
plt.plot(df['team2_runs'].tail(74),color='red',label='team2 runs')
plt.xlabel('match index')
plt.ylabel('runs')
plt.title('team1 and team2 runs trend')
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,5))
match_indices=df.index[-74:]
plt.scatter(match_indices,df['team1_runs'].tail(74),color='blue',label='team1 runs')
plt.scatter(match_indices,df['team2_runs'].tail(74),color='red',label='team2 runs')
plt.xlabel('match index')
plt.ylabel('runs')
plt.title('team1 and team2 runs trend')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['team2_wickets'].tail(74),bins=10)
plt.xlabel('wickets')
plt.ylabel('number of matches')
plt.title('distribution of wickets')
plt.show()
```

```python
toss_decision=df['toss_decision'].value_counts()
plt.bar(toss_decision.index,toss_decision.values,color=['skyblue','lightcoral'])
plt.title('distribution of toss decision')
plt.xlabel('toss decision')
plt.ylabel('number of matches')
plt.show()
```

```python
toss_decision.index
toss_decision.values
```

```python
plt.figure(figsize=(10,5))
winning_team=df['winner'].value_counts()
plt.bar(winning_team.index,winning_team.values,color=['skyblue','lightcoral'])
plt.title('distribution of winning team')
plt.xlabel('winning team')
plt.ylabel('number of matches')
plt.xticks(rotation=90)
plt.show()
```

```python
df['winner'].value_counts()
```

```python
plt.figure(figsize=(10,5))
plt.boxplot(df['team2_runs'])
plt.ylabel('team2 runs')
plt.title('team 2 runs distribution')
plt.show()
```

```python
plt.figure(figsize=(8,8))
result_type_count=df['winner'].value_counts()
plt.pie(result_type_count.values,labels=result_type_count.index,autopct='%1.1f%%',radius=2,pctdistance=1,rotatelabels=True,startangle=90)
plt.title('winner distribution')
plt.show()
```

```python
wicket_data={
    '0-3':len(df[df['team1_wickets'] <= 3]),
    '4-7':len(df[(df['team1_wickets'] <= 7)&(df['team1_wickets']>=4)]),
    '8+':len(df[df['team1_wickets']>=8])
}
plt.pie(wicket_data.values(),labels=wicket_data.keys(),autopct='%1.1f%%',startangle=90)
plt.show()
```
