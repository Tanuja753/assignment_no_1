```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
df=pd.read_csv("IPLPlayerStat.csv")
df.head()
```

```python
def player_form(avg):
    if avg <= 10:
        return"poor"
    elif avg <= 15:
        return"average"
    elif avg <= 20:
        return "good"
    elif avg <= 25:
        return "excellent"
    else:
        return "outstanding"
df["player_form"]=df["batting_avg"].apply(player_form)
df[["player","batting_avg","player_form"]].head()
```

```python
df["player_form"].unique()
```

```python
encoder=OrdinalEncoder(categories=[["poor","average","good","excellent","outstanding"]])
df["player_form_encoded"]=encoder.fit_transform(df[["player_form"]]).astype(int)
df[["player_form","player_form_encoded"]].tail(30)
```
