```python
data=pd.DataFrame(df)
```

```python
numeric_cols=data.select_dtypes("number")
numeric_cols.describe()
numeric_cols.mean()
numeric_cols.median()
numeric_cols.std()
sb.histplot(data["team1_runs"],bins=10).set_title("distribution of team1 runs")
plt.show()
sb.histplot(data["team2_runs"],bins=10).set_title("distribution of team2 runs")
plt.show()
sb.boxplot(data["team1_runs"])
plt.show()
plt.plot(data["team1_runs"],label="team1")
plt.plot(data["team2_runs"],label="team2")
```

```python
categorical_cols=data.select_dtypes("string")
categorical_cols["team1"].value_counts()
categorical_cols["team1"].value_counts(normalize=True)
data.groupby("team1")["team1_runs"].mean().plot(kind='bar',color="blue",label="team1")
plt.show()
data.groupby("team1")["team1_runs"].mean().plot(kind='pie',color="blue",label="team1")
```

```python
d1=data.groupby("toss_decision")
d2=data.groupby("winner")
tab=pd.crosstab(data["toss_decision"],data["winner"])
print(tab)
toss_win_match_win=data["toss_winner"]==data["winner"]
win_percent=toss_win_match_win.value_counts(normalize=True)*100
win_percent.plot(kind='bar',stacked=True)
```
